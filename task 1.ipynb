```python


import math
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn import linear_model
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
train_and_validation_data = pd.read_csv('/kaggle/input/house-prices-advanced-regression-techniques/train.csv')
X_test = pd.read_csv('/kaggle/input/house-prices-advanced-regression-techniques/test.csv')

y_train_and_valid = train_and_validation_data['SalePrice']
X_train_and_valid = train_and_validation_data.drop(columns=['SalePrice'])

X_train, X_valid, y_train, y_valid = train_test_split(X_train_and_valid, y_train_and_valid, test_size=0.2, random_state=0)
```

```python
categorical_column_names = [cname for cname in X_train.columns if         
                            X_train[cname].dtype == "object"]

numerical_column_names = [cname for cname in X_train.columns if 
                        X_train[cname].dtype in ['int64', 'float64']]

X_train_categorical = X_train[categorical_column_names].copy()
X_train_numerical = X_train[numerical_column_names].copy()

X_valid_categorical = X_valid[categorical_column_names].copy()
X_valid_numerical =  X_valid[numerical_column_names].copy()

X_test_categorical = X_test[categorical_column_names].copy()
X_test_numerical =  X_test[numerical_column_names].copy()
```

```python
categorical_imputer = SimpleImputer(strategy="most_frequent")

categorical_imputer.fit(X_train_categorical)     
X_train_categorical = pd.DataFrame(categorical_imputer.transform(X_train_categorical)) 
X_train_categorical.columns = categorical_column_names    
X_train_categorical.index = X_train.index


X_valid_categorical = pd.DataFrame(categorical_imputer.transform(X_valid_categorical)) 
X_valid_categorical.columns = categorical_column_names    
X_valid_categorical.index = X_valid.index

X_test_categorical = pd.DataFrame(categorical_imputer.transform(X_test_categorical)) 
X_test_categorical.columns = categorical_column_names   
X_test_categorical.index = X_test.index
```

```python
OH_encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

OH_encoder.fit(X_train_categorical)

X_train_categorical = pd.DataFrame(OH_encoder.transform(X_train_categorical))
X_train_categorical.index = X_train.index

X_valid_categorical = pd.DataFrame(OH_encoder.transform(X_valid_categorical))
X_valid_categorical.index = X_valid.index

X_test_categorical = pd.DataFrame(OH_encoder.transform(X_test_categorical))
X_test_categorical.index = X_test.index
```

```python
numerical_imputer = SimpleImputer(strategy="mean") 

numerical_imputer.fit(X_train_numerical)

X_train_numerical = pd.DataFrame(numerical_imputer.transform(X_train_numerical)) 
X_train_numerical.columns = numerical_column_names   
X_train_numerical.index = X_train.index


X_valid_numerical = pd.DataFrame(numerical_imputer.transform(X_valid_numerical))  
X_valid_numerical.columns = numerical_column_names   
X_valid_numerical.index = X_valid.index

X_test_numerical = pd.DataFrame(numerical_imputer.transform(X_test_numerical))  
X_test_numerical.columns = numerical_column_names    
X_test_numerical.index = X_test.index
```

```python
scaler = StandardScaler()

scaler.fit(X_train_numerical)

X_train_numerical = pd.DataFrame(scaler.transform(X_train_numerical))  
X_train_numerical.columns = numerical_column_names    
X_train_numerical.index = X_train.index


X_valid_numerical = pd.DataFrame(scaler.transform(X_valid_numerical))  
X_valid_numerical.columns = numerical_column_names    
X_valid_numerical.index = X_valid.index

X_test_numerical = pd.DataFrame(scaler.transform(X_test_numerical))  
X_test_numerical.columns = numerical_column_names    
X_test_numerical.index = X_test.index
```

```python
X_train = pd.concat([X_train_categorical, X_train_numerical], axis=1)
X_valid = pd.concat([X_valid_categorical, X_valid_numerical], axis=1)
X_test = pd.concat([X_test_categorical, X_test_numerical], axis=1)

X_train.columns = X_train.columns.astype(str)
X_valid.columns = X_valid.columns.astype(str)
X_test.columns = X_test.columns.astype(str)
```

```python
# linear model
regr = linear_model.LinearRegression()
```

```python
regr.fit(X_train, y_train)
```

```python
r2_train = regr.score(X_train,y_train)
print("R^2 on training data: " + str(r2_train))
```

```python
r2_validation = regr.score(X_valid,y_valid)
print("R^2 on validation data: " + str(r2_validation))
```
